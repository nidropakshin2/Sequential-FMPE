```python
config = {
    "prior":
        {
        "gamma_range": [0.05, 0.5],
        "beta_range": [0.5, 5.0],
        },
    "simulator":
        {
        "S": [1500, 2000], # int, tuple, number of Susceptible individuals or bounds for following uniform distribution
        "I": [150, 200], # int, tuple, number of Infectious individuals or bounds for following uniform distribution
        "R": [0, 50], # int, tuple, number of Recovered individuals or bounds for following uniform distribution
        "T": [100, 120],  # int, tuple, simulation duration in days or bounds for following uniform distribution
        },
    "summary": 
        {
        "type": "handmade",
        "input_dim": 3,
        "hidden_dim": 32,
        "num_layers": 2,
        "output_dim": 5,
        "dropout": 0.1,
        },
    "logger":
        {
        "name": "SIR",
        "level": "DEBUG",
        "log_file_path": "logs/sir.log",
        }
}
```

```python
import torch
from sfmpe.tasks.SIR import SIRTask


device = 'cuda' if torch.cuda.is_available() else 'cpu'
task = SIRTask(config, device)
task.summary.to(device)

theta_0 = torch.tensor([[1.2621, 0.4558]]).to(device) 
x_0 = task.summarize(task.simulate(theta_0))
```

```python
from sbi.inference import SMCABC
from matplotlib import pyplot as plt
from sfmpe.utils.plotting import dist_plot

smc = SMCABC(simulator=lambda theta: task.summary(task.simulator.simulate(theta)),
             prior=task.prior)

dist_labels=["prior", "posterior"]
param_labels=["beta", "gamma"]
stats = []


num_particles=1000
num_initial_pop=5000
num_simulations=50000
epsilon_decay=0.3


output = smc(x_0, 
    num_particles=1000,
    num_initial_pop=num_initial_pop,
    num_simulations=num_simulations,
    epsilon_decay=epsilon_decay).unsqueeze(1)
print(output.mean(dim=0), output.std(dim=0))
print("initial_pop, num_simulations, eps: ", num_initial_pop, num_simulations, epsilon_decay)
dist_plot(task.prior.sample((1000, 1, )), output, 
            true_param=theta_0,
            pred_param=output.mean(dim=0), 
            dist_labels=dist_labels, 
            param_labels=param_labels, path=f"./output/sir_smc_abc_{num_initial_pop}_{num_simulations}_{epsilon_decay}.png")
```

```python
from sbi.inference import SNPE, infer, simulate_for_sbi
from sbi.simulators.linear_gaussian import (
    linear_gaussian, 
    diagonal_linear_gaussian, 
    true_posterior_linear_gaussian_mvn_prior
)
from sbi.utils.user_input_checks import (
    process_prior,
    process_simulator,
)

theta_dim = 3
x_dim = theta_dim

prior_mean = torch.zeros(theta_dim)
prior_cov = 2*torch.eye(theta_dim)

# likelihood_mean will be likelihood_shift+theta
likelihood_shift = -1.0 * torch.zeros(x_dim)
likelihood_cov = 0.3 * torch.eye(x_dim)


num_rounds = 3
prior = torch.distributions.MultivariateNormal(
    loc=prior_mean, covariance_matrix=prior_cov
)
prior, _, prior_returns_numpy = process_prior(prior)
simulator = process_simulator(
    lambda theta: linear_gaussian(theta, likelihood_shift, likelihood_cov),
    prior,
    prior_returns_numpy,
)
theta_0, x_0 = simulate_for_sbi(simulator, proposal=prior, num_simulations=1)

inference = SNPE(prior)

proposal = prior

for _ in range(num_rounds):
    theta, x = simulate_for_sbi(simulator, proposal=proposal, num_simulations=5000)

    # In `SNLE` and `SNRE`, you should not pass the `proposal` to `.append_simulations()`.
    density_estimator = inference.append_simulations(
        theta, x, proposal=proposal
    ).train(show_train_summary=True)
    posterior = inference.build_posterior(density_estimator)
    proposal = posterior.set_default_x(x_0)
```

```python

def get_true_posterior_samples(x_o, num_samples=1):
    return true_posterior_linear_gaussian_mvn_prior(
        x_o, likelihood_shift, likelihood_cov, prior_mean, prior_cov
    ).sample((num_samples,))

theta_true = get_true_posterior_samples(x_0, 100000)
dist_plot(theta_true.unsqueeze(1), proposal.sample((100000, )).unsqueeze(1), true_param=theta_0)
```
